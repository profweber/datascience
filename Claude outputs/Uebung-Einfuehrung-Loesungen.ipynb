## Aufgabe 1: Gravitationskraft

Zwei Körper mit den Massen $m_1$ und $m_2$ im Abstand $r$ ziehen sich mit der Kraft

$$F = G \cdot \frac{m_1 \cdot m_2}{r^2}$$

an, mit der Gravitationskonstante $G = 6{,}674 \cdot 10^{-11}\,\mathrm{m^3/(kg \cdot s^2)}$.

a) Berechnen Sie die Kraft zwischen Erde ($m_E = 5{,}972 \cdot 10^{24}$ kg) und Mond ($m_M = 7{,}342 \cdot 10^{22}$ kg) bei einem mittleren Abstand von 384 400 km. Legen Sie für jede Größe eine Variable an und geben Sie das Ergebnis mit einem f-String in Exponentialschreibweise mit drei Nachkommastellen aus.

b) Berechnen Sie die Gewichtskraft einer Astronautin ($m = 80$ kg) an der Erdoberfläche mit dem Gravitationsgesetz (Erdradius $R_E = 6371$ km). Vergleichen Sie mit $F = m \cdot g$ für $g = 9{,}81\,\mathrm{m/s^2}$.

c) Die Internationale Raumstation ISS fliegt in etwa $h = 400$ km Höhe. Wie groß ist die Gravitationskraft auf die Astronautin dort? Achtung: Der Abstand wird vom **Erdmittelpunkt** aus gemessen.

**Lösung:**

````python
# a) Erde und Mond
G = 6.674e-11         # Gravitationskonstante in m³/(kg·s²)
m_earth = 5.972e24    # Masse der Erde in kg
m_moon = 7.342e22     # Masse des Mondes in kg
r = 3.844e8           # Abstand Erde–Mond in m (384 400 km)

F_moon = G * m_earth * m_moon / r**2
print(f"a) Erde–Mond: F = {F_moon:.3e} N")

# b) Astronautin an der Erdoberfläche
m = 80                # Masse der Astronautin in kg
r_earth = 6.371e6     # Erdradius in m
g = 9.81              # Fallbeschleunigung in m/s²

F_ground = G * m_earth * m / r_earth**2
print(f"b) Gravitationsgesetz: {F_ground:.1f} N, mit m·g: {m * g:.1f} N")

# c) Astronautin auf der ISS
h = 4.0e5             # Flughöhe der ISS in m
F_iss = G * m_earth * m / (r_earth + h)**2   # Klammern nicht vergessen!
print(f"c) Auf der ISS: F = {F_iss:.1f} N")
````

Die beiden Werte in b) stimmen fast überein: $g = 9{,}81\,\mathrm{m/s^2}$ ist genau die Beschleunigung, die das Gravitationsgesetz an der Erdoberfläche liefert. In c) wirkt auf der ISS immer noch fast 90 % der Gewichtskraft. Astronautinnen schweben also nicht, weil die Schwerkraft fehlt, sondern weil sie sich mit der ISS im freien Fall um die Erde befinden.

## Aufgabe 4: Ganzzahl-Division und Modulo

a) Ein Dauerlauf auf dem Prüfstand dauerte 7384 Sekunden. Geben Sie die Laufzeit im Format `2 h 3 min 4 s` aus. Verwenden Sie dazu `//` und `%`.

b) Eine Welle hat sich um insgesamt 1000° gedreht. Wie viele volle Umdrehungen sind das, und in welcher Winkelstellung (zwischen 0° und 360°) steht sie danach?

**Lösung:**

````python
# a) Laufzeit umrechnen
duration = 7384                   # Laufzeit in s
hours = duration // 3600
minutes = duration % 3600 // 60   # erst den Rest nach vollen Stunden, dann volle Minuten
seconds = duration % 60
print(f"a) Laufzeit: {hours} h {minutes} min {seconds} s")

# b) Umdrehungen und Restwinkel
angle = 1000                      # Drehwinkel in Grad
print(f"b) {angle // 360} volle Umdrehungen, Endstellung {angle % 360}°")
````

`duration % 3600 // 60` funktioniert ohne Klammern, weil `%` und `//` gleichrangig sind und von links nach rechts ausgewertet werden. Mit Klammern `(duration % 3600) // 60` ist es aber besser lesbar.

## Aufgabe 5: Erst denken, dann ausführen

Wasser bei Normaldruck ist unter 0 °C gefroren, ab 100 °C siedet es. Überlegen Sie für $T = 25$, welche Werte die `print`-Anweisungen ausgeben. Prüfen Sie Ihre Vorhersage und wiederholen Sie das Ganze für $T = 100$ und $T = -5$.

```python
T = 25                  # Wassertemperatur in °C
frozen = T < 0
boiling = T >= 100
liquid = not frozen and not boiling

print(frozen, boiling, liquid)
print(liquid == (0 <= T < 100))
print(frozen or boiling)
print(not frozen and boiling)
print(not (frozen and boiling))
```

Die beiden letzten Zeilen sehen fast gleich aus. Was ist der Unterschied? Formulieren Sie beide Ausdrücke in Worten.

**Lösung:**

````python
T = 25                  # Wassertemperatur in °C
frozen = T < 0
boiling = T >= 100
liquid = not frozen and not boiling

print(frozen, boiling, liquid)
print(liquid == (0 <= T < 100))
print(frozen or boiling)
print(not frozen and boiling)
print(not (frozen and boiling))
````

Ergebnisse für alle drei Temperaturen:

| Zeile | $T = 25$ | $T = 100$ | $T = -5$ |
|---|---|---|---|
| `frozen, boiling, liquid` | `False False True` | `False True False` | `True False False` |
| `liquid == (0 <= T < 100)` | `True` | `True` | `True` |
| `frozen or boiling` | `False` | `True` | `True` |
| `not frozen and boiling` | `False` | `True` | `False` |
| `not (frozen and boiling)` | `True` | `True` | `True` |

`not` bindet **stärker** als `and` und `or`. Die vorletzte Zeile bedeutet deshalb `(not frozen) and boiling`, also „nicht gefroren und siedend“. Die letzte Zeile bedeutet „nicht gleichzeitig gefroren und siedend“ – das ist bei Wasser immer erfüllt. Merke: Im Zweifel Klammern setzen!

Die zweite Zeile ist immer `True`: Die Definition von `liquid` und der verkettete Vergleich beschreiben denselben Temperaturbereich.

## Aufgabe 6: Überwachung eines Elektromotors

Ein Motor wird über drei Messwerte überwacht:

```python
voltage = 221.5        # Netzspannung in V
winding_temp = 104.0   # Wicklungstemperatur in °C
speed = 1460           # Drehzahl in 1/min
```

Der Motor darf betrieben werden, wenn die Spannung höchstens 10 % von 230 V abweicht, die Wicklungstemperatur höchstens 120 °C beträgt und die Drehzahl zwischen 1400 und 1500 1/min liegt (jeweils einschließlich der Grenzen).

a) Legen Sie die Wahrheitswerte `voltage_ok`, `temp_ok` und `speed_ok` an. Formulieren Sie `voltage_ok` einmal mit `and` und einmal als verketteten Vergleich.

b) `motor_ok` soll genau dann `True` sein, wenn alle drei Bedingungen erfüllt sind, `shutdown` genau dann, wenn der Motor **nicht** betrieben werden darf.

c) Eine Warnung soll erscheinen, wenn der Motor zwar noch betrieben werden darf, die Wicklungstemperatur aber über 100 °C liegt **oder** die Drehzahl weniger als 20 1/min von einer Grenze entfernt ist.

d) Ändern Sie die Messwerte und testen Sie Ihre Ausdrücke, z.B. mit 260 V, 125 °C oder 1395 1/min.

e) Eine Kommilitonin schreibt `voltage_ok = voltage >= 207 and 253`. Python meldet keinen Fehler. Testen Sie den Ausdruck mit `voltage = 300`. Was ist falsch?

**Lösung:**

````python
voltage = 221.5        # Netzspannung in V
winding_temp = 104.0   # Wicklungstemperatur in °C
speed = 1460           # Drehzahl in 1/min

# a) Einzelbedingungen
voltage_ok = voltage >= 0.9 * 230 and voltage <= 1.1 * 230
voltage_ok = 0.9 * 230 <= voltage <= 1.1 * 230          # gleichwertig als Kette
temp_ok = winding_temp <= 120
speed_ok = 1400 <= speed <= 1500

# b) Gesamtzustand
motor_ok = voltage_ok and temp_ok and speed_ok
shutdown = not motor_ok

# c) Warnung
near_limit = speed < 1420 or speed > 1480
warning = motor_ok and (winding_temp > 100 or near_limit)

print(f"Spannung ok: {voltage_ok}, Temperatur ok: {temp_ok}, Drehzahl ok: {speed_ok}")
print(f"Betrieb erlaubt: {motor_ok}, Abschalten: {shutdown}, Warnung: {warning}")

# e) der Fehler
voltage = 300
print("e)", voltage >= 207 and 253)
````

Zu c): Die Klammern um `winding_temp > 100 or near_limit` sind nötig, weil `and` stärker bindet als `or`. Ohne Klammern bedeutet der Ausdruck `(motor_ok and winding_temp > 100) or near_limit` – dann gäbe es auch dann eine „Warnung“, wenn der Motor wegen der Drehzahl längst abgeschaltet werden muss.

Zu e): Rechts von `and` steht kein Vergleich, sondern nur die Zahl `253`. Python prüft also gar nicht, ob die Spannung höchstens 253 V beträgt. Statt `True` oder `False` liefert der Ausdruck hier die Zahl `253`, und die gilt in einer Bedingung als „wahr“ – auch bei 300 V! Jede Teilbedingung muss ein vollständiger Vergleich sein: `voltage >= 207 and voltage <= 253` (oder `207 <= voltage <= 253`).

## Bonusaufgabe: Wie stark ziehen sich zwei Menschen an?

Zwei Personen mit je 70 kg stehen 1 m voneinander entfernt (verwenden Sie `G` und `g` aus Aufgabe 1). Berechnen Sie die Gravitationskraft zwischen ihnen. Vergleichen Sie diese mit der Gewichtskraft einer Mücke (Masse ca. 2,5 mg) und geben Sie aus, wie viel Mal größer die Gewichtskraft der Mücke ist. Prüfen Sie außerdem mit einem Vergleich, ob die Anziehungskraft kleiner ist als die Gewichtskraft der Mücke.

**Lösung:**

````python
m_person = 70                # Masse einer Person in kg
d = 1                        # Abstand in m
m_mosquito = 2.5e-6          # Masse der Mücke in kg (2,5 mg)

F_people = G * m_person * m_person / d**2
F_mosquito = m_mosquito * g

print(f"Anziehung zwischen den Personen: {F_people:.2e} N")
print(f"Gewichtskraft der Mücke:         {F_mosquito:.2e} N")
print(f"Die Mücke ist {F_mosquito / F_people:.0f}-mal schwerer.")
print("Anziehung kleiner als Mückengewicht?", F_people < F_mosquito)
````

Die Gravitation ist im Alltag zwischen „normalen“ Massen verschwindend gering – spürbar wird sie erst bei Massen wie der eines Planeten.
