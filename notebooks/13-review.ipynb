# Review and Exam Preparation

This chapter collects the methods of the course in one map, lists the most important pitfalls and offers practice questions in the style of the exam. The exam is written, without a computer: it tests understanding, interpretation and method choice – not programming routine. Sections marked *Extension* in the chapters are not part of the exam; in the tables below they are marked (ext.).

## Which method when?

## Core concepts at a glance

| Concept | In one sentence | Chapter |
| --- | --- | --- |
| CRISP-DM | Process model: understanding → data → preparation → modelling → evaluation → deployment, iterated. | 1 |
| Statistics vs. ML | Statistics focuses on understanding and uncertainty, ML on prediction for new data; they overlap in statistical learning. | 1 |
| Robust statistics | Median, IQR and MAD are barely affected by outliers, mean and standard deviation are. | 2 |
| MCAR / MAR / MNAR | Whether missingness is random, depends on observed data, or on the missing value itself. | 2 |
| Central limit theorem | Means of many independent observations are approximately normal, with standard error $\sigma/\sqrt{n}$. | 3 |
| Weibull shape $\beta$ | $<1$ early failures, $=1$ random failures, $>1$ wear-out. | 3 |
| Confidence interval | A procedure that captures the true parameter in 95 % of repetitions. | 3 |
| p-value | Probability of data at least as extreme as observed, assuming $H_0$. | 4 |
| Effect size | Size of a difference in physical units or standard deviations; decides relevance. | 4 |
| Prediction interval | Interval for a single new observation; wider than the confidence interval of the mean. | 5 |
| Bias–variance trade-off | Too simple models underfit, too flexible models overfit; the test error is U-shaped. | 6 |
| Cross-validation | Estimate of the test error by rotating held-out folds. | 6 |
| Precision / recall | Share of alarms that are real / share of failures that are detected. | 7 |
| Bagging / boosting | Averaging decorrelated deep trees / adding shallow trees that correct errors. | 8 |
| T² / SPE | Unusual within the PCA model plane / distance from the plane. | 9 |
| Nyquist frequency | Half the sampling rate; higher frequencies alias. | 10 |
| Time series validation | Train on the past, test on the following period. | 11 |
| Backpropagation | Chain rule to compute the gradient of the loss with respect to all weights. | 12 |

## Checklist: typical pitfalls

::: {.callout-warning}
## Before you trust a result, check …
- **Data leakage:** features that contain the answer, preprocessing fitted on all data, random splits of time series, windows or parts of the same unit in training and test.
- **Wrong metric:** accuracy or ROC on imbalanced data; MAPE near zero values.
- **Significance without relevance:** tiny effects with huge $n$; dependent observations treated as independent.
- **Extrapolation:** predictions outside the range (or the combinations) of the training data.
- **Missing baseline:** no comparison with the mean, the seasonal naive forecast or a simple linear model.
- **Correlation vs. causation:** importance and correlation do not prove that a factor causes the effect.
:::

## Typical exam question types

1. **Method choice:** given a scenario, choose and justify a method and an evaluation metric.
2. **Interpretation:** read a regression table, a confusion matrix, a scree plot, a probability plot or a learning curve.
3. **Calculation by hand:** confidence interval, test statistic, Bayes' theorem, precision/recall from a confusion matrix, aliasing frequency.
4. **Code reading:** find the error in a short Python snippet.

## Practice questions

<div class="exercises">

1. A confusion matrix shows TN = 950, FP = 30, FN = 5, TP = 15. Compute accuracy, precision and recall. Is the model useful for failure detection?
2. A sensor detects 99 % of all leaks and raises a false alarm in 2 % of the leak-free tests; 0.5 % of all parts leak. What is the probability that a part with an alarm really leaks?
3. Find the error: `X = StandardScaler().fit_transform(X); X_train, X_test, y_train, y_test = train_test_split(X, y)`.
4. A t-test on 50,000 parts shows p < 0.001 for a 0.2 µm difference in mean diameter, tolerance ±20 µm. What do you tell the production manager?
5. Explain to a colleague, in five sentences, why a random train/test split is wrong for load forecasting.
6. Ten measurements of a shaft diameter give $\bar x = 25.012$ mm and $s = 0.004$ mm. Compute a 95 % confidence interval for the mean diameter ($t_{9,\,0.975} = 2.262$).
7. A Weibull analysis of pump seals gives $\beta = 0.8$. The maintenance manager wants to replace all seals preventively after 2,000 hours. What do you advise?
8. A regression of tool wear on cutting speed gives $R^2 = 0.95$ on the training data. The CV error is three times the training error. What do you conclude, and what do you do?
9. A vibration signal is sampled at 2 kHz. A gear mesh frequency of 1.6 kHz is present. Where does it appear in the spectrum?
10. In a PCA monitoring chart, a new production run exceeds the SPE limit but not the T² limit. How do you interpret this?

### Solutions

::: {.callout-caution .solution collapse="true"}
## Solution to Question 1
Accuracy = 965/1000 = 96.5 %, precision = 15/45 = 33 %, recall = 15/20 = 75 %. The trivial model "always OK" would reach 98 % accuracy, so accuracy is misleading here. The model detects three out of four failures with two false alarms per real alarm. Whether it is useful depends on the cost ratio of missed failures and false alarms – often yes.
:::

::: {.callout-caution .solution collapse="true"}
## Solution to Question 2
$$P(\text{leak} \mid \text{alarm}) = \frac{0.99 \cdot 0.005}{0.99 \cdot 0.005 + 0.02 \cdot 0.995} = \frac{0.00495}{0.00495 + 0.0199} \approx 0.20.$$
Only about one in five alarms is a real leak – the base rate effect. Alarmed parts should be re-tested before scrapping.
:::

::: {.callout-caution .solution collapse="true"}
## Solution to Question 3
The scaler is fitted on all data before the split, so means and standard deviations of the test data leak into the preprocessing. Correct: split first, then fit the scaler on the training data only – best by putting scaler and model into a `Pipeline`. (For time series, additionally `shuffle=False` or a time-based split.)
:::

::: {.callout-caution .solution collapse="true"}
## Solution to Question 4
The difference is statistically significant but practically irrelevant: 0.2 µm is 1 % of the tolerance half-width. With 50,000 parts, the test detects even tiny differences. No action is needed because of this difference; what matters is whether the process stays within tolerance (process capability).
:::

::: {.callout-caution .solution collapse="true"}
## Solution to Question 5
Neighbouring hours and days are strongly correlated. With a random split, the test days are surrounded by training days from before *and after*, so the model only interpolates. In operation, a forecast must be made without knowing the future. The random split therefore underestimates the real forecast error. A time-based split – train on the past, test on the following period – reproduces the real situation.
:::

::: {.callout-caution .solution collapse="true"}
## Solution to Question 6
$$25.012 \pm 2.262 \cdot \frac{0.004}{\sqrt{10}} = 25.012 \pm 0.0029\ \text{mm} \;\Rightarrow\; [25.0091,\ 25.0149]\ \text{mm}.$$
The interval describes the uncertainty of the *mean* diameter, not the spread of individual shafts (about ±2s = ±0.008 mm).
:::

::: {.callout-caution .solution collapse="true"}
## Solution to Question 7
$\beta < 1$ means a decreasing failure rate – early failures dominate (e.g. installation errors, defective batches). A new seal is then *more* likely to fail soon than an old one, so preventive replacement at 2,000 hours would increase the number of failures and costs. Better: investigate the cause of the early failures (installation, quality of supply) and possibly run-in tests.
:::

::: {.callout-caution .solution collapse="true"}
## Solution to Question 8
The large gap between training and cross-validation error indicates overfitting: the model is too flexible for the amount of data, and the high training $R^2$ is misleading. Actions: simpler model (fewer terms, lower polynomial degree), regularisation, more data; select the complexity by cross-validation and report the CV or test error, not the training $R^2$.
:::

::: {.callout-caution .solution collapse="true"}
## Solution to Question 9
The Nyquist frequency is 1 kHz. The 1.6 kHz component aliases to $|1.6 - 2.0| = 0.4$ kHz = 400 Hz. The spectrum shows a false peak at 400 Hz. A sampling rate above 3.2 kHz (in practice ≥ 4 kHz) with an anti-aliasing filter is needed.
:::

::: {.callout-caution .solution collapse="true"}
## Solution to Question 10
A high SPE with normal T² means that the run shows a pattern of variation that the PCA model of normal operation does not contain – a new kind of deviation, e.g. a sensor fault, a broken correlation between variables or a new fault type. The contribution plot of the SPE shows which variables cause the deviation and is the starting point for the root-cause analysis.
:::
